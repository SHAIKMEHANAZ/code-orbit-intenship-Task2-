```python
import pandas as pd
from google.colab import files

# Upload the Excel file
uploaded = files.upload()

# Get the uploaded filename
file_name = next(iter(uploaded))

# Read the Excel dataset
df = pd.read_excel(
    file_name,
    sheet_name="Sales Data with Region"
)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

# Display the first 5 rows
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nFirst 5 Rows:")
print(df.head())

print("\nDataset Information:")
df.info()
```

```python
total_sales = df["Sales_Amount_INR"].sum()

print("Total Sales: ₹", total_sales)
```

```python
total_orders = df["Sale_ID"].nunique()

average_order_value = total_sales / total_orders

print("Total Orders:", total_orders)
print("Average Order Value: ₹", round(average_order_value, 2))
```

```python
top_products = (
    df.groupby("Product_Name")["Sales_Amount_INR"]
      .sum()
      .sort_values(ascending=False)
      .head(5)
)

print("Top 5 Products by Sales:")
print(top_products)
```

```python
category_sales = (
    df.groupby("Category")["Sales_Amount_INR"]
      .sum()
      .sort_values(ascending=False)
)

print("Sales by Category:")
print(category_sales)
```

```python
region_sales = (
    df.groupby("Region")["Sales_Amount_INR"]
      .sum()
      .sort_values(ascending=False)
)

print("Sales by Region:")
print(region_sales)
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df["Month"] = df["Order_Date"].dt.to_period("M")

monthly_sales = (
    df.groupby("Month")["Sales_Amount_INR"]
      .sum()
      .sort_index()
)

print("Monthly Sales:")
print(monthly_sales)
```

```python
with pd.ExcelWriter("CodeOrbit_Task2_Analysis_Results.xlsx") as writer:

    top_products.to_frame("Total_Sales_INR").to_excel(
        writer, sheet_name="Top Products"
    )

    category_sales.to_frame("Total_Sales_INR").to_excel(
        writer, sheet_name="Category Summary"
    )

    region_sales.to_frame("Total_Sales_INR").to_excel(
        writer, sheet_name="Region Summary"
    )

    monthly_sales.to_frame("Total_Sales_INR").to_excel(
        writer, sheet_name="Monthly Summary"
    )

print("Analysis results saved successfully!")
```
